```python
sub = []
marks = []
credithour = []
grade = []
point = []
while len(sub) < 3:
    current_sub = input("Enter subject name: ")
    
    current_gpa = float(input(f"Enter GPA for {current_sub}: "))
    
    if current_gpa < 0 or current_gpa > 4:
        print("Invalid GPA. Please enter GPA between 0 and 4.")
        continue 
        
    current_credit = int(input(f"Enter credit hour for {current_sub}: "))

    if current_gpa == 4.0:
        current_point = 4.0
        current_grade = 'A'
    elif current_gpa >= 3.7:
        current_point = 3.7
        current_grade = '-A'
    elif current_gpa >= 3.33:
        current_grade = 'B+'
        current_point = 3.33
    elif current_gpa >= 3.0:
        current_grade = 'B'
        current_point = 3.0
    elif current_gpa >= 2.0:
        current_grade = 'C'
        current_point = 2.0
    else:
        current_grade = 'F'
        current_point = 0.0
    
    point.append(current_point)
    sub.append(current_sub)
    marks.append(current_gpa)
    credithour.append(current_credit)
    grade.append(current_grade)

for i in range(3):
    print(f"{sub[i]}: GPA = {marks[i]}, Grade = {grade[i]}, Credit Hours = {credithour[i]}")
total_quality_points = 0.0
total_credits = 0.0

for i in range(3):
    total_quality_points += point[i] * credithour[i]
    total_credits += credithour[i]

if total_credits > 0:
    final_gpa = total_quality_points / total_credits
    print(f"\n>> Cumulative Semester GPA: {final_gpa:.2f} <<")
else:
    print("\nNo credit hours recorded.")
```
