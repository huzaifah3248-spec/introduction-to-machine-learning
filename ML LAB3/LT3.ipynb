```python
import numpy as np

temps = np.random.randint(15, 35, size=24)
glitch_indices = np.random.choice(24, size=3, replace=False)
temps[glitch_indices] = 0

zero_mask = (temps == 0)
valid_mask = (temps != 0)

valid_mean = np.mean(temps[valid_mask])
temps[zero_mask] = int(round(valid_mean))

print(temps)
```

```python
import numpy as np

warehouse_A = np.random.randint(10, 100, size=6)
warehouse_B = np.random.randint(10, 100, size=6)

inventory_matrix = np.vstack((warehouse_A, warehouse_B))
reshaped_matrix = inventory_matrix.reshape(4, 3)
extracted_stock = reshaped_matrix[:, :2]

print(inventory_matrix)
print(reshaped_matrix)
print(extracted_stock)
```

```python
import numpy as np

scores = np.random.randint(0, 101, size=(10, 3))

student_averages = np.mean(scores, axis=1)
subject_highest = np.max(scores, axis=0)
failing_scores = scores[scores < 40]

print(student_averages)
print(subject_highest)
print(failing_scores)
```

```python
import numpy as np

image = np.random.uniform(0.0, 255.0, size=(100, 100))
normalized_image = image / 255.0
scaled_image = np.sqrt(normalized_image)
flattened_vector = scaled_image.reshape(-1)

print(normalized_image.shape)
print(scaled_image.shape)
print(flattened_vector.shape)
```

```python
import numpy as np
region_A = np.random.randint(10, 501, size=12)
region_B = np.random.randint(10, 501, size=12)
global_sales = np.vstack((region_A, region_B))
row_indices = np.random.choice(2, size=3)
col_indices = np.random.choice(12, size=3)
global_sales[row_indices, col_indices] = -99
missing_mask = (global_sales == -99)
valid_mask = (global_sales != -99)
valid_mean = np.mean(global_sales[valid_mask])
global_sales[missing_mask] = int(round(valid_mean))
quarterly_sales = global_sales.reshape(2, 4, 3)
flattened_sales = global_sales.reshape(-1)
total_sales_per_region = np.sum(global_sales, axis=1)
log_sales = np.log(flattened_sales)
print(quarterly_sales)
print(flattened_sales.shape)
print(total_sales_per_region)
print(log_sales)
```
