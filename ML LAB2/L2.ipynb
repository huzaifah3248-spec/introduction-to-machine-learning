```python
count = 0

while count <= 0:
    count = int(input("Enter the number of prices you want to input: "))

def apply_discount(prices):
    discounted_prices = []

    for price in prices:
        if price < 0:
            print(f"Invalid price: {price}. Please enter a positive value.")
        elif price <= 100:
            discounted_prices.append(price * 0.95)
        else:
            discounted_prices.append(price * 0.80)

    return discounted_prices

prices = []
for i in range(count):
    input_price = float(input(f"Enter price {i + 1}: "))
    prices.append(input_price)

print(f"Discounted prices: {apply_discount(prices)}")
```

```python
inventory = {"Laptop": 12,"Mouse": 3,"Keyboard": 4, "Monitor": 8, "USB Cable": 2}

def check_stock(inventory):
    total_items = 0
    print("--- Stock Check ---")
    for item, count in inventory.items():
        if count < 5:
            print(f"Restock {item} (Current count: {count})")
        total_items += count

    print(f"\nTotal items in warehouse: {total_items}")

check_stock(inventory)
```

```python
students = [{"name": "Huzaifa", "age": 20, "score": 92},{"name": "Ali", "age": 21, "score": 78},{"name": "Sara", "age": 19, "score": 85},
            {"name": "Usman", "age": 22, "score": 64}]

print("Students who scored above 80:")
for student in students:
    if student["score"] > 80:
        print(f"- {student['name']} (Score: {student['score']})")

total_score = 0
for student in students:
    total_score += student["score"]

average_score = total_score / len(students)
print(f"\nAverage Score: {average_score:.2f}")

highest_student = students[0]
for student in students:
    if student["score"] > highest_student["score"]:
        highest_student = student

print(f"Top Scorer: {highest_student['name']} with {highest_student['score']} marks")

for student in students:
    if student["score"] >= 80:
        student["grade"] = "A"
    elif student["score"] >= 70:
        student["grade"] = "B"
    else:
        student["grade"] = "C"

print("\nUpdated Student Records:")
for student in students:
    print(student)
```

```python
class_data = [
    {"name": "Huzaifa", "grades": [92, 95, 88]},{"name": "Ayesha", "grades": [75, 80, 72]},
    {"name": "Bilal", "grades": [55, 62, 58]},{"name": "Zainab", "grades": [94, 91, 96]}]

def get_average(grades_list):
    if len(grades_list) == 0:
        return 0.0
    return sum(grades_list) / len(grades_list)

def process_class(class_data):
    for student in class_data:
        avg = get_average(student["grades"])
        student["average"] = avg

        if avg > 90:
            student["status"] = "Honor Roll"
        elif avg > 60:
            student["status"] = "Passing"
        else:
            student["status"] = "Failing"

process_class(class_data)

print(f"{'Name':<12} {'Average':<10} {'Status'}")
print("-" * 35)
for student in class_data:
    print(f"{student['name']:<12} {student['average']:<10.2f} {student['status']}")
```
