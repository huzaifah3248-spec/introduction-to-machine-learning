```python
count = 0

while count <= 0:
    count = int(input("Enter the number of prices you want to input: "))

def apply_discount(prices):
    discounted_prices = []

    for price in prices:
        if price < 0:
            print(f"Invalid price: {price}. Please enter a positive value.")
        elif price <= 100:
            discounted_prices.append(price * 0.95)
        else:
            discounted_prices.append(price * 0.80)

    return discounted_prices

prices = []
for i in range(count):
    input_price = float(input(f"Enter price {i + 1}: "))
    prices.append(input_price)

print(f"Discounted prices: {apply_discount(prices)}")
```
