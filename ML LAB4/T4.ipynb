```python
import pandas as pd

df = pd.read_csv("tmdb_5000_movies.csv")
df.info()
missing_percentages = (df.isnull().sum() / len(df)) * 100
highest_missing_col = missing_percentages.idxmax()
print(f"\nColumn with highest percentage of missing data: {highest_missing_col} ({missing_percentages.max():.2f}%)")

print("\nBudget Column Summary:")
print(df["budget"].describe())

memory_details = df.memory_usage(deep=True)
overview_mem = memory_details["overview"]
budget_mem = memory_details["budget"]
print(f"\nMemory used by 'overview': {overview_mem:,} bytes")
print(f"Memory used by 'budget':   {budget_mem:,} bytes")
print(f"'overview' consumes {overview_mem / budget_mem:.2f}x more memory than 'budget'")
```

```python
import pandas as pd

df = pd.read_csv("tmdb_5000_movies.csv")

hidden_gems = df[(df["vote_average"] > 8.0) & (df["popularity"] < 20)]
print(f"Total 'Hidden Gems' found: {len(hidden_gems)}")

target_languages = ["hi", "ko", "ja"]
multi_lang_movies = df[df["original_language"].isin(target_languages)]
print(f"Total Hindi, Korean, and Japanese movies: {len(multi_lang_movies)}")

flops = df[df["revenue"] < df["budget"]]
flop_percentage = (len(flops) / len(df)) * 100
print(f"Total 'Flop' movies: {len(flops)} ({flop_percentage:.2f}% of total dataset)")
```

```python
import pandas as pd

df = pd.read_csv("tmdb_5000_movies.csv")
mean_runtime = df["runtime"].mean()
df["runtime"] = df["runtime"].fillna(mean_runtime)
df["tagline"] = df["tagline"].fillna("No Tagline")

df = df.drop(columns=["homepage", "keywords"])

df = df.set_index("id")
movie_title_19995 = df.loc[19995, "original_title"]
print(f"Original title for Movie ID 19995: {movie_title_19995}")

valid_budget_median = df[df["budget"] > 0]["budget"].median()
df["budget"] = df["budget"].replace(0, valid_budget_median)
print(f"Imputed zero budgets with non-zero median: {valid_budget_median:,.2f}")
```
